```python
rng = np.random.default_rng(42)
```

```python
w = rng.dirichlet(np.ones(10))
print("Random weights:", w)
np.isclose(w.sum(), 1)
```

```python
rets = []
volas = []
shrp = []

for i in range(5000):
    w = rng.dirichlet(np.ones(10))
    rets.append(portfolio_return(w, mu))
    volas.append(portfolio_vol(w, cov))
    shrp.append(sharpe_ratio(w, mu, cov))
```

```python
from src.optimise import min_variance, max_sharpe, efficient_portfolio
```

```python
w_minvar = min_variance(cov)
w_maxsharpe = max_sharpe(mu, cov)
pd.Series(w_minvar, index=returns.columns).round(3)
print("Minimum variance portfolio return:", portfolio_return(w_minvar, mu))
print("Minimum variance portfolio volatility:", portfolio_vol(w_minvar, cov))
print("Minimum variance portfolio Sharpe ratio:", sharpe_ratio(w_minvar, mu, cov))
print("Maximum Sharpe ratio portfolio return:", portfolio_return(w_maxsharpe, mu))
print("Maximum Sharpe ratio portfolio volatility:", portfolio_vol(w_maxsharpe, cov))
print("Maximum Sharpe ratio portfolio Sharpe ratio:", sharpe_ratio(w_maxsharpe, mu, cov))
sharpe_ratio(w_maxsharpe, mu, cov) >= max(shrp)
```

```python
pd.Series(w_maxsharpe, index=returns.columns).round(3)
```

```python
targets = np.linspace(portfolio_return(w_minvar, mu), mu.max() - 0.001, 50)

frontier_vols = []

for target in targets:
    w = efficient_portfolio(mu, cov, target)
    frontier_vols.append(portfolio_vol(w, cov))

len(frontier_vols)
```

```python
plt.scatter(volas, rets, c=shrp, cmap="viridis", s=5)
plt.colorbar(label="Sharpe ratio")
plt.xlabel("Volatility(annualised)")
plt.ylabel("Return(annualised)")
plt.title("Random Portfolios: Return vs Volatility")
plt.scatter(portfolio_vol(w_equal, cov), portfolio_return(w_equal, mu), color="red", s=100, label="Equal-weighted portfolio")
plt.scatter(portfolio_vol(w_minvar, cov), portfolio_return(w_minvar, mu),
            color="green", marker="*", s=250, label="Minimum variance")
plt.scatter(portfolio_vol(w_maxsharpe, cov), portfolio_return(w_maxsharpe, mu),
            color="orange", marker="*", s=250, label="Max Sharpe")
plt.plot(frontier_vols, targets, color="black", linewidth=2, label="Efficient frontier")
plt.legend()
plt.savefig("../data/monte_carlo.png", dpi=150)
plt.show()
```
